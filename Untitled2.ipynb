```python
df_order['order_date'] = pd.to_datetime(df_order['order_date'], format='mixed')
```

```python
df_order.head()
```

```python
kategori_anomali = ['Not Available', 'unknown']
df_order['ship_mode'] = df_order['ship_mode'].replace(kategori_anomali, 'Standard Class')
```

```python
df_order['ship_mode'].value_counts()
```

```python
df_order['order_date'].dtype
```

```python
df_order['ship_mode'] = df_order['ship_mode'].fillna('Standard Class')
```

```python
df_order['ship_mode'].value_counts()
```

```python
df_order.info()
```

```python
df_order['net_sales'] = df_order['list_price'] * df_order['quantity'] * (1 - (df_order['discount_percent'] / 100))
```

```python
df_order['total_cost'] = df_order['cost_price'] * df_order['quantity']
```

```python
df_order['gross_profit'] = df_order['net_sales'] - df_order['total_cost']
```

```python
df_order['profit_margin_pct'] = (df_order['gross_profit'] / df_order['net_sales']) * 100
```

```python
df_order.head().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_category = df_order.groupby('category')[['net_sales', 'gross_profit']].sum().reset_index()
```

```python
df_category = df_category.sort_values(by='net_sales', ascending=False)
```

```python
df_melt = df_category.melt(id_vars='category', 
                             value_vars=['net_sales', 'gross_profit'], 
                             var_name='Metrik', 
                             value_name='Nilai_USD')
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
```

```python
plot = sns.barplot(
    data=df_melt, 
    x='category', 
    y='Nilai_USD', 
    hue='Metrik', 
    palette=['#2c3e50', '#e74c3c']
)
```
